## C. Daya Cleaning & Preparation

```python
df_clean = df.copy()

print(f"Dataset Shape: {df_clean.shape}")
```

#### C.1. Handling Missing Values

```python
df_clean = df_clean.dropna(subset=['Financial Stress'])

print(df_clean.isnull().sum())

print(f"\nDataset Shape: {df_clean.shape}")
```

> The missing data represents only 0.01% of the total dataset (3 out of 27,901). By removing these few rows, we ensure that all subsequent calculations, especially the custom Stress Index—are based entirely on original, authentic student responses without needing to "guess" or impute values. This maintains high data integrity for the analysis.

#### C.2. Data Type Conversion (Sleep Duration)

```python
sleep_mapping = {
    '5-6 hours': 5.5,
    '7-8 hours': 7.5,
    'More than 8 hours': 9.0,
    'Less than 5 hours': 4.0
}
df_clean['Sleep Duration'] = df_clean['Sleep Duration'].map(sleep_mapping)
```

> The Sleep Duration column is currently an "object" (text) type. You cannot calculate a correlation or a mean for a string like "5-6 hours".  Converting this to a numerical format is essential for analysis. This allows us to use NumPy for statistical comparisons and Matplotlib to create boxplots that show the actual distribution of sleep hours among students.

#### C.3. Standardizing Column Names

```python
df_clean.columns = [
    'id', 'gender', 'age', 'city', 'profession', 'academic_pressure',
    'work_pressure', 'cgpa', 'study_satisfaction', 'job_satisfaction',
    'sleep_duration', 'dietary_habits', 'degree', 'suicidal_thoughts',
    'work_study_hours', 'financial_stress', 'family_history', 'depression'
]
```

> The original column names include spaces, special characters (ex: ?) and inconsistent capitalization. This makes the code harder to write and more prone to syntax errors.
>
> This step improves professionalism and code readability, which is a requirement for a high score in the "Code Quality" section of the marking scheme. It also simplifies the process of calling columns during the visualization and analysis phases.

#### C.4. Standardizing Column Names

```python
# z score
age_mean = np.mean(df_clean['age'])
age_std = np.std(df['age'])
df_clean['age_zscore'] = (df_clean['age'] - age_mean) / age_std

outliers = df_clean[df_clean['age_zscore'].abs() > 3]

print(f"Number of outliers in Age: {len(outliers)}")
```

> The Age column may contain extreme values (outliers) that could skew the analysis of "typical" student experiences.
>
> We used NumPy to calculate Z-scores and identify rows that are more than 3 standard deviations from the mean.

## D. Feature Engineering

#### D.1. Stress Index

```python
stress_columns = ['academic_pressure', 'work_pressure', 'financial_stress']

df_clean['stress_index'] = np.mean(df_clean[stress_columns], axis=1)
```

> A student might have low academic pressure but high financial stress. The stress_index gives us a "whole-picture" view of the student's burden which is essential for answering Q3.

#### D.2. Lifestyle Score

```python
df_clean['lifestyle_score'] = (df_clean['sleep_duration'] + df_clean['study_satisfaction']) / 2
```

> This helps us answer Q2 by creating a single metric for "lifestyle health" rather than looking at sleep and satisfaction in isolation.

#### D.3. Age Grouping

```python
bins = [18, 22, 27, 60]
labels = ['Teen (18-22)', 'Young Adult (23-27)', 'Adult (28+)']

df_clean['age_group'] = pd.cut(df_clean['age'], bins=bins, labels=labels, include_lowest=True)
```

> This allows for subgroup comparisons (e.g., Is depression more prevalent in Teens or Adults?)
